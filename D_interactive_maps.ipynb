<font size="3" color="#000000"><b>4.2 Interactive Maps</b></font>



It is important to understand the geographical distribution of music artists and where they are more popular. So, this section will explore this distribution through interactive visualizations. 
Two interactive world maps will be created to understand how streaming popularity is distributed across different countries and locations. 
One map will be a world map. In this map, each artist is represented by a marker sized according to their total streams. Each marker will also be complemented with the artist’s name, origin, total streams, and percentages of collaborative and solo streams. 
And the other map will be a density map, which highlights the areas with a higher concentration of streams. 
Both of these maps will be constructed using the cleaned dataset and some information from Wikipedia.  


```python
#import libraries 
import pandas as pd
import plotly.express as px
```

```python
#import the geographic informations dataset
artists_loc= pd.read_csv('artist_streams_with_locations.csv')
artists_loc.head()
```

```python
artists_loc[['Latitude', 'Longitude']].isnull().sum()
```

We have 46 missing values in these columns. We cannot put this information in the maps. 

```python
artists_map = artists_loc.dropna(subset=['Latitude', 'Longitude'])
artists_map.shape
```

This means that we have 454 artists, with available information, and 21 columns with informations.

This is the section of the Interactive World Map. This map, as mentioned before, represents the geographical distribution of the music artists based on their origin. Each marker represents an artist, and has some information  attached about them, and its size varies according to the total streams. 

To get to the percentage of solo and collaborative streams, we need to create two new columns, and calculate them. 

```python
artists_map = artists_map.copy()

artists_map['Solo (%)'] = (artists_map['Solo Streams'] / artists_map['Total Streams'] * 100).round(2)

artists_map['Collaborative (%)'] = (artists_map['Collaborative Streams'] / artists_map['Total Streams'] * 100).round(2)
```

```python
fig = px.scatter_geo(artists_map,lat='Latitude',lon='Longitude',size='Total Streams',hover_name='Artist Name',hover_data=['Birthplace', 'Total Streams','Solo (%)', 'Collaborative (%)'])
fig.update_traces(marker_color='deeppink')
fig.update_layout(height=650,hoverlabel=dict(bgcolor='deeppink',font_color='white',font_size=13))
fig.show()
```

The World Map shows us that music artists are in many different areas of the globe, however, there is a larger number of markers in North America and Europe.
The magnitude of every single marker signifies how many overall stream counts there are for that artist. Markers that have larger sizes represent artists that generate more stream counts than those that have smaller marker sizes, which signify artists who generate less stream count.
The largest markers represent some of the most popular musicians in the dataset, including Drake, Taylor Swift, and Bad Bunny.
Another important feature of the map is its interactivity. We can find out about the artists' names, where they were born, what their total streams are, as well as the percent of solo streams and collaborative streams by simply placing our mouse over any of the markers on the map. This will help us to make comparisons among the different artists and see which type of streaming success each artist has had (i.e., whether the majority of their streaming success was due to solo songs or collaborations).
North America and Europe account for many well-known artists, but successful performers also come from other regions. Streaming success is not confined to one part of the world.
The map makes these geographic patterns easier to see and lets users examine each artist’s streaming performance individually. North America and Europe account for many well-known artists, but successful performers also come from other regions. Streaming success is not confined to one part of the world.



 Interactive Density Map

This density map highlights the geographical areas with higher concentrations of music streams, based on the artists’ birthplaces. With this map, we can identify geographical patterns and concentrations of streaming.  
If the map presents a more intense colour, it indicates a higher concentration of streams. If it presents a less intense colour, it indicates a lower concentration.


```python
fig_density = px.density_map(
    artists_map,
    lat='Latitude',
    lon='Longitude',
    z='Total Streams',
    radius=20,
    zoom=0.5,
    center={'lat': 15, 'lon': 0},
    map_style='carto-positron',
    color_continuous_scale='RdPu'
)

fig_density.update_layout(
    height=650,
)

fig_density.show()
```

The analysis of this map shows that most of the streams are concentrated in North America and parts of Europe, particularly the United States and the United Kingdom. These regions stand out because the streams are clustered there. Darker areas indicate places associated with more popular artists. By comparison, lower stream concentration is associated with artists born in Africa, South America, and Oceania. Artists from North America and Europe account for a large share of the most-streamed artists in the data. That doesn’t mean artists from other regions aren’t successful; the data covers only a specific group of well-known artists. Streaming popularity varies by region, with some areas contributing far more to the total number of streams than others. Artists from those regions are likely contributing more to the total numbers. Keep in mind that the map shows where artists were born, not where their listeners are. Also, 46 artists are missing because their locations couldn’t be found, which may affect the overall picture.

<font size="4" color="#43A832"><b>Battle of the Stars</b></font>


This section takes a more interactive approach to understanding what helps a music artist succeed.
Rather than counting streams alone, we will rank artists across several measures, including overall totals, solo work, collaborations, and performance over their careers.
Each category will be ranked separately, so we can see which artists perform best under each measure of success and how their positions shift from one set of criteria to another.
Users can also choose artists to compare, with interactive visuals showing where they excel and how they differ.
Musical success has no single measure: an artist may lead one category but not necessarily perform as well in another.


```python
# Identify the 10 artists with more streams 
# Identify the 10 artists with the most streams
top_streams = artists_loc.sort_values('Total Streams', ascending=False).head(10)
top_streams[['Artist Name', 'Total Streams']]
```

```python
# Visual ranking´
fig_top = px.bar(top_streams,x='Total Streams',y='Artist Name',orientation='h',color='Total Streams',color_continuous_scale='RdPu',title='Top 10 Artists')
fig_top.update_layout(height=550,yaxis=dict(autorange='reversed'))

fig_top.show()
```

Drake leads the ranking with 137,492.1 million total streams. Taylor Swift follows at 127,861.0 million, while Bad Bunny has 125,899.8 million.
These three artists sit well above the rest. The gap is especially clear beside the fourth-ranked artist, The Weeknd, with 95,703.2 million streams.
Streaming popularity is heavily concentrated at the top: the first three artists record much higher totals than everyone else in the ranking. This category counts total streams alone. It says nothing about how long an artist has worked or how much collaborations matter.
The next categories use other measures of success to see whether these artists still lead the rankings.



<font size="4" color="#43A832"><b>2. Streams per Career Year</b></font>


```python
artists_games = artists_loc.copy()
artists_games['Career Years'] = (artists_games['Year'] - artists_games['Debut Year'] + 1)
artists_games[['Artist Name', 'Debut Year', 'Year', 'Career Years']].head(10)
```

```python
artists_games['Streams per Career Year'] = (artists_games['Total Streams'] / artists_games['Career Years'])
top_career = artists_games.sort_values('Streams per Career Year', ascending=False).head(10)
top_career[['Artist Name', 'Career Years', 'Streams per Career Year']]
```

```python
fig_career = px.bar(
    top_career,
    x='Streams per Career Year',
    y='Artist Name',
    orientation='h',
    color='Streams per Career Year',
    color_continuous_scale='RdPu',
    title='The Fastest Riser — Top 10 Artists'
)

fig_career.update_layout(
    height=550,
    yaxis=dict(autorange='reversed')
)

fig_career.show()
```

If we compare this ranking to the previous one, we can see that Bad Bunny jumps from third to first place, with around 8,992.8 millions of streams per career year.
Drake moves to the second place, with 6,547.2 millions, followed by Taylor Swift, with 6,088.6 millions.
This change happens because Bad Bunny has gained a very large number of streams in a smaller career. While Drake and Taylor Swift have been active for 21 years, Bad Bunny has only recorded 14 career years according to the dataset.
Another interesting point is the presence of artists like Billie Eilish, Olivia Rodriao and Peso Pluma in the Top 10. These artists were not included in the ranking, but their shorter career show them to reach higher values when streams are divided by career length. 
These results show us that the way we measure success can dramatically change the rankings. Artists with longer careers may have accumulated more total streams, while artists with shorter careers can stand out when we consider streams per career year.
However, this is just an approximation.



<font size="4" color="#43A832"><b>3. The Solo Champion</b></font>


```python
# Calculate the percentage of solo streams
artists_games['Solo Percentage'] = (artists_games['Solo Streams'] / artists_games['Total Streams'] * 100).round(2)
artists_games[['Artist Name', 'Total Streams', 'Solo Streams', 'Solo Percentage']].head(10)
```

```python
top_solo = artists_games.sort_values(['Solo Percentage', 'Total Streams'],ascending=[False, False]).head(10)
top_solo[['Artist Name', 'Total Streams', 'Solo Percentage']]
```

```python
fig_solo = px.bar(
    top_solo,
    x='Total Streams',
    y='Artist Name',
    orientation='h',
    color='Total Streams',
    color_continuous_scale='RdPu',
    title='The Solo Champion — Top 10 Artists'
)

fig_solo.update_layout(
    height=550,
    yaxis=dict(autorange='reversed')
)

fig_solo.show()
```

As you can see all the artists in this table have 100% of their streams coming from their solo work. We have Harry Styles on the top of the table with 29,449 million streams, AC/DC with 17,010 million streams and Nirvana with 16,228 million streams.
What is interesting to see is that in this table we can find a wide range of artists. Beside modern artists we can find famous artists and bands such as ABBA, Led Zeppelin, Eagles and The Smiths.
Unlike the previous categories in this table what matters more to an artist is to have more streams coming from their solo work compared to their popularity. Since all the artists have the same percentage, their total number of streams was used to determine their positions in the ranking.
As the result, we can see that there are artists that are more successful through their solo work. However, this does not mean that those artists have more streams than those who have more streams coming from their collaborations.



<font size="4" color="#43A832"><b>4. The Collaboration Kings</b></font>


```python
artists_games['Collaborative Percentage'] = (artists_games['Collaborative Streams'] /artists_games['Total Streams'] * 100).round(2)
artists_games[['Artist Name', 'Collaborative Percentage']].head(10)
```

```python
top_collab = artists_games.sort_values(['Collaborative Percentage', 'Total Streams'],ascending=[False, False]).head(10)
top_collab[['Artist Name', 'Total Streams', 'Collaborative Percentage']]
```

```python
fig_collab = px.bar(
    top_collab,
    x='Total Streams',
    y='Artist Name',
    orientation='h',
    color='Total Streams',
    color_continuous_scale='RdPu',
    title='The Collaboration Kings — Top 10 Artists'
)

fig_collab.update_layout(
    height=550,
    yaxis=dict(autorange='reversed')
)

fig_collab.show()
```

Disney leads the ranking with 100% of its streams coming from collaborations.

Amitabh Bhattacharya and Irshad Kamil are tied with Disney with 100% of their streams coming from collaborations. Some of the names you may not recognize internationally are Irshad Kamil, Kumaar and Amitabh Bhattacharya.
Ranked ninth in the top 10 of artists who have the highest percentage of streams coming from collaborations with the biggest amount of streams in this group, but not 100% ( 99.96%) , is Pritam.
This is the percentage of streams coming from collaborations versus the total amount of streams. This is not a measure of overall popularity. Disney has the highest percentage of streams coming from collaborations 100%, but this does not necessarily make it the most popular artist overall.

<font size="4" color="#43A832"><b>6. Interactive Artist Battle</b></font>


```python
artist1 = 'Drake'
artist2 = 'Taylor Swift'

battle = artists_games[(artists_games['Artist Name'] == artist1) |(artists_games['Artist Name'] == artist2)]
battle[['Artist Name', 'Total Streams', 'Solo Percentage',
        'Collaborative Percentage', 'Streams per Career Year']]
```

```python
fig_battle = px.bar(
    battle,
    x='Artist Name',
    y=['Solo Percentage', 'Collaborative Percentage'],
    barmode='group',
    color_discrete_sequence=['deeppink', 'purple'],
    title='Artist Battle — Solo vs Collaborative Streams'
)

fig_battle.update_layout(
    height=500,
    yaxis_title='Percentage (%)'
)

fig_battle.show()
```

```python
import ipywidgets as widgets
```

```python
artist_options = sorted(artists_games['Artist Name'].dropna().unique())

artist1_select = widgets.Dropdown(
    options=artist_options,
    value='Drake',
    description='Artist 1:'
)

artist2_select = widgets.Dropdown(
    options=artist_options,
    value='Taylor Swift',
    description='Artist 2:'
)

display(artist1_select, artist2_select)
```

```python
def artist_battle(artist1, artist2):
    battle = artists_games[
        (artists_games['Artist Name'] == artist1) |
        (artists_games['Artist Name'] == artist2)
    ]

    fig = px.bar(
        battle,
        x='Artist Name',
        y=['Solo Percentage', 'Collaborative Percentage'],
        barmode='group',
        color_discrete_sequence=['deeppink', 'purple'],
        title='Artist Battle — Solo vs Collaborative Streams'
    )

    fig.update_layout(
        height=500,
        yaxis_title='Percentage (%)'
    )

    fig.show()
```

```python
battle_interactive = widgets.interactive_output(
    artist_battle,
    {
        'artist1': artist1_select,
        'artist2': artist2_select
    }
)

display(battle_interactive)
```

```python
def full_artist_battle(artist1, artist2):

    battle = artists_games[
        (artists_games['Artist Name'] == artist1) |
        (artists_games['Artist Name'] == artist2)
    ]

    # Compare solo and collaborative percentages
    fig1 = px.bar(
        battle,
        x='Artist Name',
        y=['Solo Percentage', 'Collaborative Percentage'],
        barmode='group',
        color_discrete_sequence=['deeppink', 'purple'],
        title='Artist Battle — Solo vs Collaborative Streams'
    )

    fig1.update_layout(
        height=500,
        yaxis_title='Percentage (%)',
        legend_title='Stream Type'
    )

    fig1.show()

    # Compare overall and career performance
    fig2 = px.bar(
        battle,
        x='Artist Name',
        y=['Total Streams', 'Streams per Career Year'],
        barmode='group',
        color_discrete_sequence=['purple', 'deeppink'],
        title='Overall Performance vs Career Performance'
    )

    fig2.update_layout(
        height=500,
        yaxis_title='Streams (millions)',
        legend_title='Performance Measure'
    )

    fig2.show()
```

```python
full_battle_output = widgets.interactive_output(
    full_artist_battle,
    {
        'artist1': artist1_select,
        'artist2': artist2_select
    }
)

display(full_battle_output)
```

```python
def artist_winners(artist1, artist2):

    battle = artists_games[
        (artists_games['Artist Name'] == artist1) |
        (artists_games['Artist Name'] == artist2)
    ]

    metrics = [
        'Total Streams',
        'Streams per Career Year',
        'Solo Percentage',
        'Collaborative Percentage'
    ]

    for metric in metrics:

        artist1_value = battle[battle['Artist Name'] == artist1][metric].iloc[0]
        artist2_value = battle[battle['Artist Name'] == artist2][metric].iloc[0]

        if artist1_value > artist2_value:
            print(metric, ':', artist1, 'wins!')

        elif artist2_value > artist1_value:
            print(metric, ':', artist2, 'wins!')

        else:
            print(metric, ': It is a tie!')
```

```python
winner_output = widgets.interactive_output(
    artist_winners,
    {
        'artist1': artist1_select,
        'artist2': artist2_select
    }
)

display(winner_output)
```

This interactive compares musical performance, as measured by streams, of two artists. Musical success can be measured in many different ways.
In addition to looking at the total number of streams, you can examine streams per year in a career, streams per year in a solo career, and streams per year in a collaborative career.
One can see a difference in the amount of success different artists have by looking at measurements other than total number of streams. Looking into other measurements can show which artists perform better than others.
